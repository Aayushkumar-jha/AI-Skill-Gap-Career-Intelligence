# Phase 3: Data Cleaning & Exploratory Data Analysis (EDA)
## AI Skill-Gap & Career Intelligence Engine

This notebook demonstrates data acquisition, cleaning, deduplication, and market demand analytics across 12,000+ job postings.

```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import config

conn = sqlite3.connect(config.DB_PATH)
df_jobs = pd.read_sql_query('SELECT * FROM jobs', conn)
df_skills = pd.read_sql_query('SELECT * FROM skills', conn)
conn.close()

print(f'Loaded {len(df_jobs)} jobs and {len(df_skills)} standardized skills.')
df_jobs.head(3)
```

### 1. Missing Value and Cleanliness Audit

```python
print('Missing values in Job Postings:')
print(df_jobs.isnull().sum())

print('\nJob Distribution across 12 Target Roles:')
print(df_jobs['role_category'].value_counts())
```

### 2. Salary Distribution Across Roles & Experience Tiers

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_jobs, x='role_category', y='salary_lpa', hue='experience_tier')
plt.xticks(rotation=45, ha='right')
plt.title('Industry Compensation Distribution (LPA) by Role & Seniority')
plt.ylabel('Salary (INR LPA)')
plt.tight_layout()
plt.show()
```
