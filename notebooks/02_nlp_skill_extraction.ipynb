# Phase 4: NLP Skill Extraction & Candidate Profiling
## AI Skill-Gap & Career Intelligence Engine

Demonstration of multi-tier NLP parsing: canonical alias matching, regex boundary chunking, and contextual proficiency estimation.

```python
from src.nlp.skill_extractor import SkillExtractor
from src.nlp.resume_parser import ResumeParser

extractor = SkillExtractor()

sample_resume = '''
Senior Machine Learning Engineer with 4 years experience.
Expert in Python, PyTorch, Scikit-Learn, and Docker.
Built and deployed LLM solutions using Transformers, LangChain (RAG), and FastAPI.
Experienced in SQL and PostgreSQL query tuning.
'''

extracted_skills = extractor.extract_skills_with_proficiency(sample_resume)
exp_years = ResumeParser.extract_experience_years(sample_resume)
edu = ResumeParser.extract_education(sample_resume)

print(f'Detected Experience: {exp_years} Years')
print(f'Detected Education: {edu}')
print(f'Extracted {len(extracted_skills)} Skills with Proficiency Confidence:')
for s, p in sorted(extracted_skills.items(), key=lambda x: x[1], reverse=True):
    print(f'  - {s:35s}: {p:.2f}')
```
