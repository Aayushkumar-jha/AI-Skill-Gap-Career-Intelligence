# Phase 8 & 11: Role Classification, Explainability & What-If Simulation
## AI Skill-Gap & Career Intelligence Engine

Covers feature vector transformation, Random Forest model inference, explainable AI (XAI) feature attribution, and What-If simulation.

```python
from src.features.feature_engineering import FeatureEngineer
from src.models.role_classifier import RoleClassifier
from src.explainability.shap_analysis import ExplainabilityEngine

fe = FeatureEngineer()
clf = RoleClassifier()
explainer = ExplainabilityEngine(clf, fe.get_feature_names())

cand_skills = {
    'Python': 0.90, 'SQL': 0.85, 'Pandas & NumPy': 0.85, 'Scikit-Learn': 0.80,
    'Exploratory Data Analysis': 0.80, 'Probability & Statistics': 0.75
}

feat_vec = fe.candidate_to_feature_vector(cand_skills, 2.0, "Bachelor's")
pred_df = clf.predict_roles(feat_vec)
print('Role Suitability Predictions:')
print(pred_df.head(5).to_string(index=False))
```

### SHAP Feature Attribution (Explainable AI)

```python
explanation = explainer.explain_candidate_prediction(feat_vec, 'Data Scientist')
print('\nTop Positive Factors Boosting Score:')
for pos in explanation['positive_drivers']:
    print(f"  + {pos['feature']}: {pos['contribution']:+.3f}")

print('\nTop Missing Gap Penalties:')
for neg in explanation['negative_penalties']:
    print(f"  - {neg['feature']}: {neg['contribution']:+.3f}")
```
